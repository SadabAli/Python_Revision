DataFrame.append() is deprecated (removed in newer pandas).

So this pattern is:

❌ slow

❌ deprecated

❌ will break in new pandas versions

Why append() was removed (important concept)

append() creates a new DataFrame every time

This causes O(n²) time complexity

Very inefficient in loops

concat():

Combines everything once

Much faster

More memory efficient

```python
# loop on groupby
df = pd.concat(
    [data[data['IMDB_Rating'] == data['IMDB_Rating'].max()] for _, data in genres], # List comprihention
    ignore_index=True
)
```

```python
# split (apply) combine
# apply -> builtin function

genres.apply(min)# enres{split}.apply{apply}(min){combine}
```

```python
# find number of movies starting with A for each group
genres['Series_Title'].str.startswith('A').sum()
```

```python
def A(dataset):
    return dataset['Series_Title'].str.startswith('A').sum()

genres.apply(A)
```

```python
# find normalized IMDB rating group wise

def normal(group):
  group['norm_rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
  return group

genres.apply(normal)
```
