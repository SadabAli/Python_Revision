```python
# GroupBy Attributes and Methods
# find total number of groups -> len
# find items in each group -> size
# first()/last() -> nth item
# get_group -> vs filtering
# groups
# describe
# sample
# nunique
```

```python
# find total number of groups -> len
len(movies.groupby('Genre'))
```

```python
movies['Genre'].unique().size
```

```python
# find items in each group -> size
movies.groupby('Genre').size()
```

```python
# first()/last() -> nth item

movies.groupby('Genre').first() # har genre ka 1st movie print hoga e.g. 14
```

```python
movies.groupby('Genre').last()# har genre ka last movie print hoga
```

```python
genres = movies.groupby('Genre')
genres.nth(6) # 6th meas 5th
```

```python
# get_group -> vs filtering
genres.get_group('Horror')
```

```python
movies[movies['Genre'] == 'Horror'] #filtering long process
```

```python
# describe
genres.describe()
```

```python
genres.sample(3,replace=True) # har genre se 3 random value display karega
```
