### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
import pandas as pd
import numpy as np
```

### Series from list

```python
# string 
country = ['india','russia','Brazil']
pd.Series(country)
```

```python
# integer 
runs = [12,43,23,44,33]
pd.Series(runs)
```

```python
#custom index 
marks=[100,100,99,98]
subjests=['ml','ai','genai','math']

pd.Series(marks,index=subjests)
```

```python
# setting a name
marks = pd.Series(marks,index=subjests,name='Shadab ke marks')
marks
```

### Series from dictionary

```python
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'hindi':100
}

marks_series = pd.Series(marks,name='sadab ke marks')
marks_series
```
