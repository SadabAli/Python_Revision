What happens inside __init__?

Every time you create a new object, Python runs __init__.

Inside it, you set:

self.customer_id = 0
self.customer_id = self.customer_id + 1 (start with 1)

So for EVERY object:

customer_id starts at 0
then becomes 1

That’s why every object gets 1.

```python
## ATM code

class Atm:
    __counter=1 # static variable
    def __init__(self):
        self.__balance = 0 # instnve variabl (obj.<variable_name>)
        self.pin=''
        self.customer_id=Atm.__counter
        Atm.__counter += 1 # static variable (class_name.<variable_name>)
        # self.menu()

    @staticmethod # this is utility function / decorator
    def get_counter():
        return Atm.__counter

    def get_balance(self):
        return self.__balance
    
    def set_balance(self, new_value):
        if type(new_value) == int:
            self.__balance = new_value
        else:
            print("fuck you!!!!!!")

    def menu(self):

        user_input=input(
            '''1 for create pin
            2 for check balance
            3 for withdrowl
            4 for exit'''
        )

        if user_input == '1':
            self.create_pin()
        elif user_input == '2':
            self.check_balance()
        elif user_input =='3':
            self.withdrowl()
        else:
            exit()
    def create_pin(self):
        user_pin=input("Enter your pin")
        self.pin=user_pin
        user_balance=int(input("Enter balance"))
        self.__balance=user_balance

        print("Pint successfully created")

        # self.menu()

    def check_balance(self):
        user_pin=input("Enter your pin")

        if user_pin == self.pin :
            print("Your amount is ", self.__balance)

        # self.menu()

    def withdrowl(self):
        user_pin=input("Enter your pin")
        
        if user_pin == self.pin:
            amount=int(input("Enter amount to be withdrowl"))
            if amount <= self.__balance :
                self.__balance = self.__balance - amount
                print(" withdrowl successfull")

                print("balance is", self.__balance)

            else:
                print('chalna chutiye')
        else:
            print('Wrong pin')

        # self.menu()
```
