```python
# WAP for table and input takes forom user

number=int(input('Enter a number'))

for i in range(1,11):
    print( number,"x",i,"=",i*number)
```

```python
# using while loop
number=int(input('Enter a number'))

i=1
while i<11:
    print( number,"x",i,"=",i*number)
    i +=1
```
