```python
# return even and odd from a list

l=[1,2,3,4,5,6,7,8,9]

odd=[i for i in l if i%2!=0]
even=[i for i in l if i%2==0]
print(even)
print(odd)
```

```python
#exeption handling

def add(a,b):
    try:
        if isinstance(a,int) and isinstance(b,int):
            return a+b
        else:
            raise ValueError("Both value should be Integer")
    except Exeption as r:
        print('Error',r)

add('r',2)
```
