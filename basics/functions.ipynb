```python
# args and kwargs
# used to take infinite input

# functions for addition win n number of input

def Add(*number):
    """
    This function is used for addition and can take infinite inputs
    
    """
    result = 0

    for i in number:
        result += i
    return result

Add(1,2,3,4,5)
```

```python
# kwargs meas key-wors arguments

def dictt(**dictionary):
    for (keys,values) in dictionary.items():
        print(keys,':',values)
dictt(name='sadab', age=21, status='Bhadwaa')
```

```python
# function without return statements

def is_even(num):
    """
    When a function has no return statement, Python automatically returns None
    
    """
    if num%2 == 0:
        print('even')
    else:
        print('odd')
is_even(6)

```

## Difference between print and return

print = shouting the answer

Anyone can hear it, but you cannot use it again.

return = giving the answer on paper

You can store it, use it, send it, calculate with it.

```python
# using print
def add(a, b):
    print(a + b)

result = add(5, 3) # automatically return None
print("Result =", result) 
```

```python
# using return
def add(a, b):
    return a + b

result = add(5, 3)
print("Result =", result)
```

## what happens if a function has multiple returns

Python executes ONLY the first return that it reaches

```python
def test():
    return 1
    return 2
    return 3

print(test())
```

```python
def check_number(n):
    if n > 0:
        return "positive"
    elif n < 0:
        return "negative"
    else:
        return "zero"
check_number(242)
```

## What is a Higher-Order Function?


A higher-order function is a function that takes another function as an argument or returns a function


Examples of higher-order functions:
map(), filter(), sorted(), reduce(), and even your own custom functions.

```python
# HOF

def transform(functionn,listt):
    output=[]
    for i in listt:
        output.append(functionn(i))
    print(output)

listt=[1,2,3,4]
transform(lambda x:x**2,listt)
```

```python
# map (map(function/lambda, iterable))

# apply any logic on the list, basically it create new list from given list based on logic

square = list(map(lambda x:x**2, [1,2,3]))
square
```

```python
l=[1,2,3,4]
even = list(map(lambda x:'even' if x%2==0 else 'odd',l))
even
```

```python
users=[
    {
        'name':'sadab',
        'age':21
    },
    
    {
        'name':'simpal',
        'age':24
    }
]

users
```

```python
list(map(lambda users:users['name'],users))
```

```python
# filter- used to filter from the list based on condition
# syntax = filter(function, iterable)
l=[1,2,3,4,5,6,7]

num=list(filter(lambda x:x>3, l))
num
```

```python
l=['apple','banana','patato','bringel']

fruits=list(filter(lambda x:x.startswith(('a','b')), l))
fruits
```

```python
# separate fruits from a mixed list of fruits + vegetables.

items = ['apple', 'banana', 'potato', 'brinjal', 'orange', 'tomato']
fruits = ['apple', 'banana', 'orange', 'mango', 'grapes']

fruit=list(filter(lambda x: x in fruits,items))
fruit
```

```python
# reduce - used to give answer in single values based on condition

from functools import reduce

nums = [1, 2, 3, 4, 5]
result = reduce(lambda x, y: x + y, nums)
print(result)
```
