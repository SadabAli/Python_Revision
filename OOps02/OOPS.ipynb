```python
## ATM code

class Atm:
    def __init__(self):
        self.balance = 0
        self.pin=''
        self.menu()

    def menu(self):

        user_input=input(
            '''1 for create pin
            2 for check balance
            3 for withdrowl
            4 for exit'''
        )

        if user_input == '1':
            self.create_pin()
        elif user_input == '2':
            self.check_balance()
        elif user_input =='3':
            self.withdrowl()
        else:
            exit()
    def create_pin(self):
        user_pin=input("Enter your pin")
        self.pin=user_pin
        user_balance=int(input("Enter balance"))
        self.balance=user_balance

        print("Pint successfully created")

        self.menu()

    def check_balance(self):
        user_pin=input("Enter your pin")

        if user_pin == self.pin :
            print("Your amount is ", self.balance)

        self.menu()

    def withdrowl(self):
        user_pin=input("Enter your pin")
        
        if user_pin == self.pin:
            amount=int(input("Enter amount to be withdrowl"))
            if amount <= self.balance :
                self.balance = self.balance - amount
                print(" withdrowl successfull")

                print("balance is", self.balance)

            else:
                print('chalna chutiye')
        else:
            print('Wrong pin')

        self.menu()





```
