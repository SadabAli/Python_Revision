```python
# factorial

nu=int(input('Enter a number'))

fact = 1
for i in range(1,nu+1):
    fact *= i
print(fact)
```

```python
# Prime no.

def IsPrime(number):
    if number <= 1:
        return "Not a prime number"
    
    for i in range(2, number):
        if number % i == 0:
            return "Not a prime number"
    
    return "Prime number"

print(IsPrime(3))
```

```python
# fide prime in between 1-20qs

for number in range(2,21):
    for i in range(2,number):
        if number%i == 0:
            break
    else:
        print(i,'Prime number')
```

```python
# check the s string is palindrome or not
def Palindrome(s):
    if s == s[::-1]:
        print("Palindrome")
    else:
        return 'not a palindrome'
Palindrome('sadab')
```
