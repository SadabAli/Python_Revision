```python
# A decorator must:

# Take a function as input

# Define an inner (wrapper) function

# Return the wrapper function

def design(fun):
    def wrapper():
        print('******************')
        fun()
        print('******************')
    return wrapper

def display():
    print('Hy i am Tejaswini')

def dis():
    print('hy i am sadab')

d=design(display)
d()
s=design(dis)
s()
```

```python
def design(fun):
    def wrapper():
        print('******************')
        fun()
        print('******************')
    return wrapper
@design
def display():
    print('Hy i am Tejaswini')
display()
```

```python
# this is NOT a decorator
# design() calls the function immediately

# It does not return a function

# No wrapping happens

# No function modification is saved

def design(fun):
    
    print('******************')
    fun()
    print('******************')
   

def display():
    print('Hy i am Tejaswini')

def dis():
    print('hy i am sadab')

design(display)
design(dis)
```
