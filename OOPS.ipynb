```python
# pass by reference

class Me:
    def __init__(self,name,age):
        self.name = name 
        self.age = age

def user_name(user):
    print("my name is ",user.name,'and i am ',user.age,' years old')

obj=Me('simpal',20)
user_name(obj)

# here a function take input as a class
```

```python
class Me:
    def __init__(self,name,age):
        self.name = name 
        self.age = age

def user_name(user):
    print(id(user))
    print("my name is ",user.name,'and i am ',user.age,' years old')

obj=Me('simpal',20)
print(id(obj))
user_name(obj)

# tech. i dont give an object i give refrence of an objact (pass by reference)
```

```python
# Object is mutable/changebla

class Me:
    def __init__(self,name,age):
        self.name = name 
        self.age = age

def user_name(user):
    print("my name is ",user.name,'and i am ',user.age,' years old')
    user.name='tejaswini'

obj=Me('simpal',20)
user_name(obj)
print(obj.name)
```

```python
class Me:
    def __init__(self,name,age):
        self.name = name 
        self.age = age

def user_name(user):
    user.name='tejaswini'
    return user # without return, in python it default return None not user reference

obj=Me('simpal',20)
print(id(obj))
obj2=user_name(obj)
print(id(obj2))

```

What happens here?

obj is created → stored in memory

You print its id → that is its memory address

obj is sent into the function user_name

Inside the function, you only change the name attribute

You return the same object

obj2 receives the same object reference

```python
# encapsulation

# instance variable

class abc:
    def __init__(self,name):
        self.name=name # self.name → instance variable (acted diff for diff obj)

obj1=abc('sadab')
obj2=abc('simpal')
print(obj1.name)
print(obj2.name)
```
