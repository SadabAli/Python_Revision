### Stacking

```python
# horizontal stacking
a4 = np.arange(12).reshape(3,4)
a5 = np.arange(12,24).reshape(3,4)
print(a4)
print(a4.shape)
decoratorr()
print(a5)
print(a5.shape)
```

```python
s=np.hstack((a4,a5))
print(s)
print(s.shape)
```

```python
t=np.vstack((a4,a5))
print(t)
print(t.shape)
```
