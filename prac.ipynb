```python
# @MethodOverRidding

class Phone:
    def __init__(self):
        pass 
    def buy(self):
        print('Inside the phone') 
class Smartphone(Phone):
    def __init__(self):
        pass
     
    def buy(self):
        print('I miss you simpal') 

s=Smartphone() 
s.buy()
```

```python
def add(*num):
    result=0 
    for i in num:
        result +=i 
    return result 
add(1,2,3)
```
